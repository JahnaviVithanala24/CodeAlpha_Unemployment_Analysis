```python
average_unemployment = df['Estimated Unemployment Rate (%)'].mean()

print("Average Unemployment Rate:",
      round(average_unemployment, 2), "%")
```

```python
highest = df.loc[
    df['Estimated Unemployment Rate (%)'].idxmax()
]

highest
```

```python
print("Region:", highest['Region'])
print("Date:", highest['Date'])
print("Unemployment Rate:",
      highest['Estimated Unemployment Rate (%)'])
```

```python
lowest = df.loc[
    df['Estimated Unemployment Rate (%)'].idxmin()
]

lowest
```

```python
monthly_unemployment = df.groupby('Date')[
    'Estimated Unemployment Rate (%)'
].mean()
```

```python
monthly_unemployment
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    monthly_unemployment.index,
    monthly_unemployment.values,
    marker='o'
)

plt.title("Average Unemployment Rate Over Time")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
region_unemployment = df.groupby('Region')[
    'Estimated Unemployment Rate (%)'
].mean().sort_values(ascending=False)

region_unemployment
```

```python
plt.figure(figsize=(12,8))

region_unemployment.plot(kind='bar')

plt.title("Average Unemployment Rate by Region")
plt.xlabel("Region")
plt.ylabel("Average Unemployment Rate (%)")
plt.xticks(rotation=90)

plt.show()
```

```python
top_10 = region_unemployment.head(10)

top_10
```

```python
pre_covid = df[
    df['Date'] < '2020-03-01'
]

covid_period = df[
    df['Date'] >= '2020-03-01'
]
```

```python
print(
    "Pre-COVID Average:",
    round(
        pre_covid['Estimated Unemployment Rate (%)'].mean(),
        2
    )
)

print(
    "COVID Period Average:",
    round(
        covid_period['Estimated Unemployment Rate (%)'].mean(),
        2
    )
)
```

```python
pre_covid_avg = pre_covid[
    'Estimated Unemployment Rate (%)'
].mean()

covid_avg = covid_period[
    'Estimated Unemployment Rate (%)'
].mean()

change = covid_avg - pre_covid_avg

print("Change in unemployment rate:",
      round(change, 2), "percentage points")
```

```python
plt.figure(figsize=(12,6))

plt.plot(
    monthly_unemployment.index,
    monthly_unemployment.values,
    marker='o'
)

plt.axvline(
    pd.Timestamp('2020-03-01'),
    linestyle='--',
    label='COVID-19 Period'
)

plt.title("Unemployment Rate Trend and COVID-19 Period")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")

plt.legend()
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
df['Month'] = df['Date'].dt.month
```

```python
monthly_pattern = df.groupby('Month')[
    'Estimated Unemployment Rate (%)'
].mean()

monthly_pattern
```

```python
average_employed = df['Estimated Employed'].mean()

print(
    "Average Estimated Employed:",
    round(average_employed, 2)
)
```

```python
employment_trend = df.groupby('Date')[
    'Estimated Employed'
].mean()
```

```python
avg_participation = df[
    'Estimated Labour Participation Rate (%)'
].mean()

print(
    "Average Labour Participation Rate:",
    round(avg_participation, 2),
    "%"
)
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x='Estimated Labour Participation Rate (%)',
    y='Estimated Unemployment Rate (%)'
)

plt.title(
    "Labour Participation Rate vs Unemployment Rate"
)

plt.show()
```

```python
numeric_columns = [
    'Estimated Unemployment Rate (%)',
    'Estimated Employed',
    'Estimated Labour Participation Rate (%)'
]

correlation = df[numeric_columns].corr()

correlation
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Matrix")

plt.show()
```
